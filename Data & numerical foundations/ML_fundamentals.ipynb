```python
import numpy as np


def gradient_descent():
    
    w = 10
    learning_rate = 0.1
    
    for i in range(20):
        gradient = 2 * w
        w = w - (learning_rate * gradient)
        
        print(f"Epoch: {i+1}, New: {w:.3f} ")
        
gradient_descent()
```

```python
import numpy as np

# Linear regretion from scratch
# Training dataset

X = np.array([1,2,3,4,5], dtype=float) # samples
Y = np.array([2,4,6,8,10], dtype=float) # Labels

# Initial parameters

w = 0.0
b= 0.0

learning_rate = 0.01
epochs = 1000

for epoch in range(epochs):
    # Find initial prediction
    y_pred = w * X + b
    
    # Find loss
    loss = np.mean((Y - y_pred)**2)
    
    # Find gradients
    dw = np.mean((-2*X) * (Y - y_pred))
    db = np.mean(-2 * (Y - y_pred))
    
    # Update parameters
    
    w = w - learning_rate * dw
    b = b - learning_rate * db
    
    if epoch % 100 == 0:
        print(f"Epoch: {epoch:03}  w: {w:.3f}  b: {b:.3f}  Loss: {loss:.3f}")
        
    
    
```
