```python
import numpy as np

# An aoutlier is a datapoint that is unusually different from others 
# Ex. in [1,2,3,4,5,6,7,8,50] 50 is an aoutlier

# We can detect outliers using diferent ways like using graphical method and using IQR method

data = np.array([10, 11, 12, 13, 14, 15, 16, 18, 20, 100,-10]) # 100

np.sort(data)

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

print("Q1: ", Q1)
print("Q3: ", Q3)

IQR = Q3 - Q1

print("IQR: ", IQR)

lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

# Boolean indexing 

print(data < lower_limit)
print(data > upper_limit)

print((data > upper_limit) | (data < lower_limit))

outliers = data[(data > upper_limit) | (data < lower_limit)]

print("The outliers are: ", outlier)
```
