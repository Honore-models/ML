```python
import numpy as np

# ML functions

x = np.array([1,2,3,4,5])

linear = 2*x + 4
quadratic = x**2 + 6
exponential = np.exp(x)
logarithmic = np.log(x)
sigmoid = 1 / (1 + np.exp(-x))
ReLU = np.maximum(0,x)

print("linear: \n", linear)
print("quadratic: \n", quadratic)
print("exponential: \n", exponential)
print("logarithmic: \n", logarithmic)
print("Sigmoid: \n", sigmoid)
print("ReLU: \n", ReLU)
```
