```python
import numpy as np

# Solving systems of linear equation AX = B
# Where A is the coeficient matrix and B is the equal values matrix

A = np.array([
    [1,1,1],
    [2,-1,1],
    [1,2,-1]
])

B = np.array([6,2,7])

np.linalg.solve(A,B)
Det = np.linalg.det(A)
print(f"The determinant of A is {Det:.2f}")

np.hstack((A,B.reshape(3,1)))
```

```python
import numpy as np

A = np.array([
    [4,-3,1],
    [2,1,3],
    [-1,2,-5]
])

b = np.array([-10,0,17])
np.shape(A)
np.linalg.solve(A,b)
det = np.linalg.det(A)

print(f"The determinant of\n {A} is \n{det:.2f}")
```
